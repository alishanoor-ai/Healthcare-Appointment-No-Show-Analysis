```python
# Load dataset
df = pd.read_csv('KaggleV2-May-2016.csv')

# Display first 5 rows
df.head()
```

```python
# Check dataset shape
print("Rows and Columns:", df.shape)

# Check column names
print("\nColumn Names:")
print(df.columns.tolist())

# Dataset information
df.info()
```

```python
# First 5 rows
display(df.head())

# Last 5 rows
display(df.tail())

# Statistical summary
display(df.describe())

# Missing values
print("Missing Values:")
print(df.isnull().sum())

# Duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())

# Unique values in important columns
print("\nGender:", df['Gender'].unique())
print("No-show:", df['No-show'].unique())
print("Age range:", df['Age'].min(), "to", df['Age'].max())
```

```python
# Check negative ages
print("Negative Age Records:")
print(df[df['Age'] < 0].shape[0])

# Check ages above 100
print("\nAge Above 100:")
print(df[df['Age'] > 100].shape[0])

# Check age frequency
print("\nAge Value Counts:")
print(df['Age'].value_counts().sort_index())
```

```python
# Display records with negative age or age above 100
unusual_age = df[(df['Age'] < 0) | (df['Age'] > 100)]

display(unusual_age)
```

```python
# Count each unusual age
print(unusual_age['Age'].value_counts().sort_index())
```

```python
# Remove records with negative age
df = df[df['Age'] >= 0].copy()

# Check updated dataset shape
print("Updated Dataset Shape:", df.shape)

# Verify negative ages
print("Negative Age Records:", (df['Age'] < 0).sum())
```

```python
# Convert date columns into datetime format

df['ScheduledDay'] = pd.to_datetime(df['ScheduledDay'])
df['AppointmentDay'] = pd.to_datetime(df['AppointmentDay'])

# Check updated data types

print(df[['ScheduledDay', 'AppointmentDay']].dtypes)
```

```python
# Check first 5 date records

display(df[['ScheduledDay', 'AppointmentDay']].head())
```

```python
# Calculate waiting days between scheduling and appointment

df['WaitingDays'] = (
    df['AppointmentDay'].dt.normalize() -
    df['ScheduledDay'].dt.normalize()
).dt.days

# Check results
print("Waiting Days Summary:")
print(df['WaitingDays'].describe())

# Check negative waiting days
print("\nNegative Waiting Days:", (df['WaitingDays'] < 0).sum())

# Display first 10 records
display(df[['ScheduledDay', 'AppointmentDay', 'WaitingDays']].head(10))
```

```python
# Find records with negative waiting days

negative_waiting = df[df['WaitingDays'] < 0]

# Display those records
display(negative_waiting[
    ['PatientId', 'ScheduledDay', 'AppointmentDay', 'WaitingDays']
])

# Count negative records
print("Negative Waiting Records:", len(negative_waiting))
```

```python
# Remove records with negative waiting days

df = df[df['WaitingDays'] >= 0].copy()

# Check updated dataset
print("Updated Dataset Shape:", df.shape)

# Verify negative waiting days
print("Negative Waiting Days:", (df['WaitingDays'] < 0).sum())
```

```python
# Check unique values in categorical columns

columns = [
    'Gender',
    'No-show',
    'Scholarship',
    'Hipertension',
    'Diabetes',
    'Alcoholism',
    'Handcap',
    'SMS_received'
]

for col in columns:
    print(f"\n{col}:")
    print(df[col].value_counts())
```

```python
# Check handicap values and percentages

print("Handcap Value Counts:")
print(df['Handcap'].value_counts().sort_index())

print("\nHandcap Percentages:")
print(df['Handcap'].value_counts(normalize=True).sort_index() * 100)
```

```python
# Age distribution summary

print("Age Statistics:")
print(df['Age'].describe())

# Count patients by age group

df['AgeGroup'] = pd.cut(
    df['Age'],
    bins=[-1, 12, 18, 35, 50, 65, 120],
    labels=['Children', 'Teenagers', 'Young Adults',
            'Adults', 'Middle Age', 'Seniors']
)

print("\nAge Group Distribution:")
print(df['AgeGroup'].value_counts())
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x='AgeGroup',
    order=df['AgeGroup'].value_counts().index
)

plt.title('Patient Distribution by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Number of Patients')
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

```python
# Count appointments by No-show status

no_show_counts = df['No-show'].value_counts()

print("Appointment Attendance:")
print(no_show_counts)

# Calculate percentages

no_show_percentages = df['No-show'].value_counts(normalize=True) * 100

print("\nAttendance Percentages:")
print(no_show_percentages)
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x='No-show',
    order=['No', 'Yes']
)

plt.title('Appointment Attendance Status')
plt.xlabel('No-show Status')
plt.ylabel('Number of Appointments')

plt.xticks([0, 1], ['Attended', 'No-show'])

plt.tight_layout()
plt.show()
```

```python
# Gender-wise appointment attendance

gender_analysis = pd.crosstab(
    df['Gender'],
    df['No-show']
)

print("Gender-wise Attendance:")
display(gender_analysis)

# Calculate percentages

gender_percentage = pd.crosstab(
    df['Gender'],
    df['No-show'],
    normalize='index'
) * 100

print("Gender-wise Attendance Percentage:")
display(gender_percentage)
```

```python
gender_percentage.plot(
    kind='bar',
    figsize=(8, 5)
)

plt.title('Gender-wise Appointment Attendance')
plt.xlabel('Gender')
plt.ylabel('Percentage (%)')
plt.xticks(rotation=0)
plt.legend(title='No-show')
plt.tight_layout()
plt.show()
```

```python
# Age group-wise appointment attendance

age_analysis = pd.crosstab(
    df['AgeGroup'],
    df['No-show']
)

print("Age Group-wise Attendance:")
display(age_analysis)

# Calculate percentages

age_percentage = pd.crosstab(
    df['AgeGroup'],
    df['No-show'],
    normalize='index'
) * 100

print("Age Group-wise Attendance Percentage:")
display(age_percentage)
```

```python
age_percentage['Yes'].sort_values(ascending=False).plot(
    kind='bar',
    figsize=(9, 5)
)

plt.title('No-show Rate by Age Group')
plt.xlabel('Age Group')
plt.ylabel('No-show Rate (%)')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# SMS reminder-wise attendance

sms_analysis = pd.crosstab(
    df['SMS_received'],
    df['No-show']
)

print("SMS-wise Attendance:")
display(sms_analysis)

# Calculate percentages

sms_percentage = pd.crosstab(
    df['SMS_received'],
    df['No-show'],
    normalize='index'
) * 100

print("SMS-wise Attendance Percentage:")
display(sms_percentage)
```

```python
sms_percentage['Yes'].plot(
    kind='bar',
    figsize=(7, 5)
)

plt.title('No-show Rate by SMS Reminder')
plt.xlabel('SMS Received (0 = No, 1 = Yes)')
plt.ylabel('No-show Rate (%)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
df['WaitingDays'].describe()
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(df['WaitingDays'], bins=40, kde=True)

plt.title('Distribution of Appointment Waiting Days')
plt.xlabel('Waiting Days')
plt.ylabel('Number of Appointments')

plt.tight_layout()
plt.show()
```

```python
waiting_analysis = df.groupby('WaitingDays')['No-show'].apply(
    lambda x: (x == 'Yes').mean() * 100
)

print(waiting_analysis.head(20))
```

```python
plt.figure(figsize=(10, 5))

waiting_analysis.plot(kind='line', marker='o')

plt.title('No-show Rate by Waiting Days')
plt.xlabel('Waiting Days')
plt.ylabel('No-show Rate (%)')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
df['AppointmentMonth'] = df['AppointmentDay'].dt.to_period('M')

monthly_appointments = df['AppointmentMonth'].value_counts().sort_index()

print(monthly_appointments)
```

```python
plt.figure(figsize=(9, 5))

monthly_appointments.plot(kind='bar')

plt.title('Monthly Appointment Trends')
plt.xlabel('Appointment Month')
plt.ylabel('Number of Appointments')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
neighbourhood_counts = df['Neighbourhood'].value_counts()

print(neighbourhood_counts.head(10))
```

```python
plt.figure(figsize=(11, 6))

neighbourhood_counts.head(10).sort_values().plot(
    kind='barh'
)

plt.title('Top 10 Neighbourhoods by Appointment Count')
plt.xlabel('Number of Appointments')
plt.ylabel('Neighbourhood')

plt.tight_layout()
plt.show()
```

```python
neighbourhood_noshow = df.groupby('Neighbourhood')['No-show'].apply(
    lambda x: (x == 'Yes').mean() * 100
)

top_noshow_areas = neighbourhood_noshow.sort_values(
    ascending=False
).head(10)

print(top_noshow_areas)
```

```python
plt.figure(figsize=(11, 6))

top_noshow_areas.sort_values().plot(kind='barh')

plt.title('Top 10 Neighbourhoods by No-show Rate')
plt.xlabel('No-show Rate (%)')
plt.ylabel('Neighbourhood')

plt.tight_layout()
plt.show()
```

```python
scholarship_analysis = df.groupby('Scholarship')['No-show'].apply(
    lambda x: (x == 'Yes').mean() * 100
)

print(scholarship_analysis)
```

```python
plt.figure(figsize=(7, 5))

scholarship_analysis.plot(kind='bar')

plt.title('No-show Rate by Scholarship')
plt.xlabel('Scholarship (0 = No, 1 = Yes)')
plt.ylabel('No-show Rate (%)')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
health_columns = ['Hipertension', 'Diabetes', 'Alcoholism']

for col in health_columns:
    result = df.groupby(col)['No-show'].apply(
        lambda x: (x == 'Yes').mean() * 100
    )

    print(f"\n{col} vs No-show:")
    print(result)
```

```python
health_data = df.groupby('No-show')[health_columns].mean()

health_percentage = df.groupby('No-show')[health_columns].apply(
    lambda x: x.mean() * 100
)

health_percentage.T.plot(
    kind='bar',
    figsize=(9, 5)
)

plt.title('Health Conditions and Appointment Attendance')
plt.xlabel('Health Conditions')
plt.ylabel('Percentage (%)')
plt.xticks(rotation=0)
plt.legend(title='No-show')

plt.tight_layout()
plt.show()
```
